## Section 1: What Is OOP and Why Should I Care?

So far you've written **procedural** code — a sequence of steps that runs top-to-bottom, organized with functions. That works great and you'll keep using it.

**Object-Oriented Programming (OOP)** is a different way to organize code. Instead of thinking "what steps do I need to do?", you think "what *things* does my program deal with, and what can those things *do*?"

### The Key Vocabulary

| Term | What It Means | Real-World Analogy |
|------|--------------|--------------------|
| **Class** | A blueprint/template that defines what something is | A cookie cutter |
| **Object** | One specific instance built from that blueprint | An actual cookie |
| **Attribute** | A piece of data that belongs to an object (like a variable) | The cookie's flavor |
| **Method** | A function that belongs to an object (an action it can perform) | The cookie's ability to crumble |

Think about a `Dog` class:
- **Attributes** (what it knows): name, breed, age
- **Methods** (what it can do): bark(), eat(), fetch()

You define the class *once*, then create as many dog objects as you want — each with its own name, breed, and age.

Let's see this in action. The example below shows the **same program** written two ways — procedural (what you know) and OOP (what we're learning).

````python
# EXAMPLE: Procedural vs OOP — the same program written two ways

# ===== PROCEDURAL (what you already know) =====
# Data is stored in separate variables
student_name = "Alice"
student_gpa = 3.8
student_grades = [92, 88, 95]

# Functions are separate — you pass data INTO them
def get_average(grades):          # takes a list of grades as input
    return sum(grades) / len(grades)

def is_honor_roll(gpa):           # takes a GPA as input
    return gpa >= 3.5

print("=== Procedural ===")
print(f"{student_name} - Average: {get_average(student_grades):.1f}")
print(f"Honor roll? {is_honor_roll(student_gpa)}")


# ===== OOP (what we're learning) =====
# Data AND functions are bundled together inside a class
class Student:                           # define a blueprint called Student
    def __init__(self, name, gpa, grades):  # constructor — runs when you create a new Student
        self.name = name                  # store name as an attribute of this object
        self.gpa = gpa                    # store gpa as an attribute
        self.grades = grades              # store grades as an attribute
    
    def get_average(self):               # method — notice 'self' is the first parameter
        return sum(self.grades) / len(self.grades)  # uses self.grades — its OWN data
    
    def is_honor_roll(self):             # another method
        return self.gpa >= 3.5           # uses self.gpa — its OWN data

# Create an object (instance) of Student
alice = Student("Alice", 3.8, [92, 88, 95])  # calls __init__ automatically

print("\n=== OOP ===")
print(f"{alice.name} - Average: {alice.get_average():.1f}")  # alice already knows her own grades
print(f"Honor roll? {alice.is_honor_roll()}")                 # alice already knows her own GPA
````

**Key takeaway:** In the procedural version, data and functions are separate — you have to pass the data in every time. In the OOP version, the object *carries its own data with it*. When you call `alice.get_average()`, you don't need to tell it which grades to use — Alice already knows her own grades.

## Section 2: Building a Class Step by Step

Here's the anatomy of a class:

```python
class ClassName:                        # 1. The class name (capitalize it)
    def __init__(self, param1, param2): # 2. The constructor (sets up a new object)
        self.attribute1 = param1        # 3. Store data as attributes using self.
        self.attribute2 = param2
    
    def some_method(self):              # 4. Methods — functions that belong to the object
        # can use self.attribute1, self.attribute2, etc.
        return self.attribute1
```

**What is `self`?** Every method's first parameter is `self`. It's a reference to *this particular object*. When you have 5 different Dog objects, `self` is how each one knows to use *its own* name and age, not some other dog's.

**What is `__init__`?** It's the **constructor** — a special method that Python calls automatically every time you create a new object. It's where you set up the object's starting data.

````python
# EXAMPLE: Building a Dog class from scratch

class Dog:
    def __init__(self, name, breed, age):  # constructor — called when you create a Dog
        self.name = name                    # store name as an attribute (self.name)
        self.breed = breed                  # store breed as an attribute (self.breed)
        self.age = age                      # store age as an attribute (self.age)
    
    def bark(self):                         # a method — an action the Dog can perform
        print(f"{self.name} says: Woof!")   # uses self.name to know ITS name
    
    def describe(self):                     # another method
        print(f"{self.name} is a {self.age}-year-old {self.breed}.")


# Create two different Dog objects from the SAME class blueprint
# Python calls __init__ automatically each time, passing in the arguments
dog1 = Dog("Buddy", "Golden Retriever", 3)   # creates one Dog with these values
dog2 = Dog("Max", "German Shepherd", 5)       # creates another Dog with different values

# Each object has its OWN data
dog1.describe()    # Buddy is a 3-year-old Golden Retriever.
dog2.describe()    # Max is a 5-year-old German Shepherd.

dog1.bark()        # Buddy says: Woof!
dog2.bark()        # Max says: Woof!

# You can read and change attributes directly with dot notation
print(f"\n{dog1.name} is {dog1.age} years old")
dog1.age = 4                     # happy birthday, Buddy!
print(f"After birthday: {dog1.name} is now {dog1.age}")
````

## Section 3: Methods That Return Values

Methods are just functions that live inside a class. Like any function, they can **return** a value that you can use elsewhere — store it in a variable, print it, use it in a calculation, etc.

The only difference from a regular function is that they take `self` as the first parameter so they can access the object's attributes.

````python
# EXAMPLE: A Rectangle class with methods that return values

class Rectangle:
    def __init__(self, width, height):
        self.width = width              # store width attribute
        self.height = height            # store height attribute
    
    def area(self):                     # method that RETURNS a value
        return self.width * self.height # calculates and returns the area
    
    def perimeter(self):               # another method that returns a value
        return 2 * (self.width + self.height)
    
    def describe(self):                # method that prints (doesn't return)
        print(f"Rectangle {self.width}x{self.height}")
        print(f"  Area: {self.area()}")       # calls its OWN area method
        print(f"  Perimeter: {self.perimeter()}")  # calls its OWN perimeter method


# Create two rectangles
r1 = Rectangle(5, 3)                  # 5 wide, 3 tall
r2 = Rectangle(10, 7)                 # 10 wide, 7 tall

r1.describe()                          # prints info about r1
r2.describe()                          # prints info about r2

# Since area() RETURNS a value, we can use it in calculations
total_area = r1.area() + r2.area()     # add the two areas together
print(f"\nTotal area of both rectangles: {total_area}")
````

## Section 4: Inheritance — Building on Existing Classes

**Inheritance** is when you create a new class that's based on an existing class. The new class automatically gets *everything* the original class has (all its attributes and methods), and you can add new stuff on top.

- **Parent class** (also called base or superclass) — the original class
- **Child class** (also called derived or subclass) — the new class that inherits

**Why is this useful?** Imagine you have an `Animal` class with basic stuff every animal has (name, speak). You can then make `Dog` and `Cat` classes that inherit all the animal stuff but each add their own unique features (dogs have tricks, cats purr).

**Syntax:** Put the parent class name in parentheses after the child class name:
```python
class Dog(Animal):    # Dog inherits from Animal
```

````python
# EXAMPLE: Inheritance — Animal parent class with Dog and Cat children

# PARENT CLASS — defines the basics that ALL animals share
class Animal:
    def __init__(self, name, sound):    # every animal has a name and a sound
        self.name = name
        self.sound = sound
    
    def speak(self):                    # every animal can speak
        print(f"{self.name} says {self.sound}!")


# CHILD CLASS — Dog inherits from Animal and adds its own features
class Dog(Animal):                      # (Animal) means Dog inherits from Animal
    def __init__(self, name):           # Dog's constructor only needs a name
        super().__init__(name, "Woof")  # super() calls the PARENT's __init__
                                         # so we don't have to rewrite that code
        self.tricks = []                 # new attribute only Dogs have (not all Animals)
    
    def learn_trick(self, trick):        # new method only Dogs have
        self.tricks.append(trick)        # add the trick to this dog's list
        print(f"{self.name} learned {trick}!")


# CHILD CLASS — Cat inherits from Animal and adds its own features
class Cat(Animal):
    def __init__(self, name):
        super().__init__(name, "Meow")  # calls Animal's __init__ with sound "Meow"
    
    def purr(self):                      # new method only Cats have
        print(f"{self.name} purrs softly...")


# Create objects
buddy = Dog("Buddy")
whiskers = Cat("Whiskers")

# Both can speak — inherited from Animal
buddy.speak()                # Buddy says Woof!
whiskers.speak()             # Whiskers says Meow!

# Only Dogs can learn tricks
buddy.learn_trick("sit")
buddy.learn_trick("shake")
print(f"{buddy.name}'s tricks: {buddy.tricks}")

# Only Cats can purr
whiskers.purr()

# This would ERROR — Cats don't have learn_trick:
# whiskers.learn_trick("roll over")   # AttributeError!
````

## Section 5: Three Programming Paradigms (Exam Vocabulary!)

The exam (competency **5.1.4**) asks you to describe, compare, and contrast three programming styles. You've now experienced two of them. Here's what you need to know:

| Paradigm | Core Idea | How You've Used It |
|----------|-----------|--------------------|
| **Procedural** | Code is a sequence of steps organized into functions. Data and functions are separate. | Everything you wrote in Python and C before today. |
| **Object-Oriented (OOP)** | Code is organized around objects that bundle data and behavior together. | What we just learned — classes, objects, methods. |
| **Event-Driven** | Code runs in response to events (clicks, keypresses, timers). The program waits for something to happen, then reacts. | Web pages (clicking a button triggers JavaScript), GUIs, games. |

**The key difference to remember:**
- **Procedural** = "Do step 1, then step 2, then step 3."
- **OOP** = "Create things that know stuff and can do stuff."
- **Event-driven** = "Wait for something to happen, then respond."

Most real programs **mix all three**. A video game might use OOP for characters (classes with attributes and methods), procedural logic inside those methods (step-by-step calculations), and event-driven code to respond to controller input.

## Quick Reference

| What You Want To Do | Syntax |
|---------------------|--------|
| Define a class | `class MyClass:` |
| Constructor | `def __init__(self, param1, param2):` |
| Store an attribute | `self.name = name` |
