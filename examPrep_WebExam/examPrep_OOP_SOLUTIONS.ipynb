## Section 1 Example: Procedural vs OOP

```python
# Procedural
student_name = "Alice"
student_gpa = 3.8
student_grades = [92, 88, 95]

def get_average(grades):
    return sum(grades) / len(grades)

def is_honor_roll(gpa):
    return gpa >= 3.5

print("=== Procedural ===")
print(f"{student_name} - Average: {get_average(student_grades):.1f}")
print(f"Honor roll? {is_honor_roll(student_gpa)}")

# OOP
class Student:
    def __init__(self, name, gpa, grades):
        self.name = name
        self.gpa = gpa
        self.grades = grades
    
    def get_average(self):
        return sum(self.grades) / len(self.grades)
    
    def is_honor_roll(self):
        return self.gpa >= 3.5

alice = Student("Alice", 3.8, [92, 88, 95])
print("\n=== OOP ===")
print(f"{alice.name} - Average: {alice.get_average():.1f}")
print(f"Honor roll? {alice.is_honor_roll()}")
```

## Section 2 Example: Building a Class

```python
class Dog:
    def __init__(self, name, breed, age):
        self.name = name
        self.breed = breed
        self.age = age
    
    def bark(self):
        print(f"{self.name} says: Woof!")
    
    def describe(self):
        print(f"{self.name} is a {self.age}-year-old {self.breed}.")

dog1 = Dog("Buddy", "Golden Retriever", 3)
dog2 = Dog("Max", "German Shepherd", 5)
dog1.describe()
dog2.describe()
dog1.bark()
dog2.bark()
print(f"\n{dog1.name} is {dog1.age} years old")
dog1.age = 4
print(f"After birthday: {dog1.name} is now {dog1.age}")
```

### Practice Task 1 — SOLUTION

```python
class Phone:
    def __init__(self, brand, model, battery):
        self.brand = brand          # store brand as attribute
        self.model = model          # store model as attribute
        self.battery = battery      # store battery level (0-100)
    
    def describe(self):
        print(f"{self.model} ({self.brand}) — Battery: {self.battery}%")
    
    def charge(self):
        self.battery = 100          # set battery to full
        print("Phone fully charged!")

# Create two phones and test them
phone1 = Phone("Apple", "iPhone 15", 85)
phone2 = Phone("Samsung", "Galaxy S24", 42)

phone1.describe()     # iPhone 15 (Apple) — Battery: 85%
phone2.describe()     # Galaxy S24 (Samsung) — Battery: 42%

phone2.charge()       # Phone fully charged!
phone2.describe()     # Galaxy S24 (Samsung) — Battery: 100%
```

## Section 3 Example: Methods That Return Values

```python
class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height
    
    def area(self):
        return self.width * self.height
    
    def perimeter(self):
        return 2 * (self.width + self.height)
    
    def describe(self):
        print(f"Rectangle {self.width}x{self.height}")
        print(f"  Area: {self.area()}")
        print(f"  Perimeter: {self.perimeter()}")

r1 = Rectangle(5, 3)
r2 = Rectangle(10, 7)
r1.describe()
r2.describe()
total_area = r1.area() + r2.area()
print(f"\nTotal area of both rectangles: {total_area}")
```

### Practice Task 2 — SOLUTION

```python
class BankAccount:
    def __init__(self, owner):
        self.owner = owner          # store the owner's name
        self.balance = 0            # every new account starts at $0
    
    def deposit(self, amount):      # takes an amount to add
        self.balance += amount      # add to balance
        print(f"Deposited ${amount:.2f}. New balance: ${self.balance:.2f}")
    
    def withdraw(self, amount):     # takes an amount to remove
        if amount > self.balance:   # check if there's enough money
            print(f"Insufficient funds! Balance: ${self.balance:.2f}")
        else:
            self.balance -= amount  # subtract from balance
            print(f"Withdrew ${amount:.2f}. New balance: ${self.balance:.2f}")
    
    def get_balance(self):          # returns the balance (doesn't print)
        return self.balance

# Test it
acct = BankAccount("Alice")
acct.deposit(100)        # Deposited $100.00. New balance: $100.00
acct.withdraw(30)        # Withdrew $30.00. New balance: $70.00
acct.withdraw(200)       # Insufficient funds! Balance: $70.00
print(f"Final balance: ${acct.get_balance():.2f}")  # Final balance: $70.00
```

## Section 4 Example: Inheritance

```python
class Animal:
    def __init__(self, name, sound):
        self.name = name
        self.sound = sound
    
    def speak(self):
        print(f"{self.name} says {self.sound}!")

class Dog(Animal):
    def __init__(self, name):
        super().__init__(name, "Woof")
        self.tricks = []
    
    def learn_trick(self, trick):
        self.tricks.append(trick)
        print(f"{self.name} learned {trick}!")

class Cat(Animal):
    def __init__(self, name):
        super().__init__(name, "Meow")
    
    def purr(self):
        print(f"{self.name} purrs softly...")

buddy = Dog("Buddy")
whiskers = Cat("Whiskers")
buddy.speak()
whiskers.speak()
buddy.learn_trick("sit")
buddy.learn_trick("shake")
print(f"{buddy.name}'s tricks: {buddy.tricks}")
whiskers.purr()
```

### Practice Task 3 — SOLUTION

```python
# Parent class
class Vehicle:
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year
    
    def describe(self):
        print(f"{self.year} {self.make} {self.model}")

# Child class — inherits from Vehicle, adds battery range
class ElectricCar(Vehicle):
    def __init__(self, make, model, year, battery_range):
        super().__init__(make, model, year)  # call parent's __init__
        self.battery_range = battery_range    # new attribute for ElectricCar only
    
    def range_info(self):                     # new method for ElectricCar only
        print(f"Range: {self.battery_range} miles on a full charge")

# Test it
my_car = ElectricCar("Tesla", "Model 3", 2024, 350)
my_car.describe()      # inherited from Vehicle: 2024 Tesla Model 3
my_car.range_info()    # ElectricCar's own method: Range: 350 miles on a full charge
```

### Practice Task 4 — SOLUTION (Vocabulary)

```python
# 1. What is a class?
# A blueprint or template that defines what attributes (data) and methods (actions) 
# objects of that type will have. You define it once, then create objects from it.

# 2. What is an object?
# A specific instance of a class. Each object has its own copy of the data defined
# by the class. For example, if Dog is a class, then Buddy and Max are objects.

# 3. What is the difference between an attribute and a method?
# An attribute is data stored in the object (like a variable — name, age, etc.)
# A method is a function that belongs to the object (an action it can do — bark, describe, etc.)

# 4. What does __init__ do?
# It's the constructor — a special method that runs automatically when you create a
# new object. It sets up the object's starting attributes.

# 5. What does 'self' refer to?
# 'self' is a reference to the current/specific object. It lets the object access
# its own attributes and methods. Like saying "my" — self.name means "my name".

# 6. What is inheritance?
# Creating a new class based on an existing class. The child class gets all the
# parent's attributes and methods, and can add its own new ones.

# 7. What is the difference between procedural and OOP?
# Procedural organizes code as a sequence of steps/functions — data and functions
# are separate. OOP organizes code around objects that bundle data and functions
# together — the object carries its own data with it.
```

### Section 6 Mini Project — SOLUTION

```python
class Character:
    def __init__(self, name, attack_power):
        self.name = name                # character's name
        self.health = 100               # everyone starts at 100 HP
        self.attack_power = attack_power # how much damage they deal
    
    def attack(self, other):            # 'other' is the character being attacked
        other.health -= self.attack_power  # reduce the OTHER character's health
        if other.health < 0:            # don't let health go below 0
            other.health = 0
        print(f"{self.name} attacks {other.name} for {self.attack_power} damage!")
    
    def is_alive(self):                 # returns True if HP > 0
        return self.health > 0
    
    def status(self):                   # print current status
        print(f"  {self.name}: {self.health} HP")

# Create two characters
hero = Character("Hero", 25)            # hero does 25 damage per attack
villain = Character("Villain", 20)      # villain does 20 damage per attack

# Fight loop — keep going until someone runs out of HP
round_num = 1
while hero.is_alive() and villain.is_alive():  # both must be alive to continue
    print(f"\n--- Round {round_num} ---")
    hero.attack(villain)                # hero attacks first
    if villain.is_alive():              # only attack back if villain survived
        villain.attack(hero)
    hero.status()                       # show both HP totals
    villain.status()
    round_num += 1                      # next round

# Announce winner
print("\n=== BATTLE OVER ===")
if hero.is_alive():
    print(f"{hero.name} wins with {hero.health} HP remaining!")
else:
    print(f"{villain.name} wins with {villain.health} HP remaining!")
```
