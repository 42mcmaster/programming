## Section 2 Example: AND Operator

```python
print(f"True  and True  = {True and True}")
print(f"True  and False = {True and False}")
print(f"False and True  = {False and True}")
print(f"False and False = {False and False}")

has_license = True
has_car = False
can_drive = has_license and has_car
print(f"\nHas license: {has_license}")
print(f"Has car: {has_car}")
print(f"Can drive? {can_drive}")
```

### Practice Task 2 — SOLUTION

```python
my_answers = [
    True,    # True and True = True (both are True)
    False,   # True and False = False (second is False)
    False,   # False and True = False (first is False)
    False    # False and False = False (both are False)
]

correct = [True and True, True and False, False and True, False and False]
labels = ["True  and True ", "True  and False", "False and True ", "False and False"]

for i in range(4):
    status = "CORRECT" if my_answers[i] == correct[i] else "WRONG"
    print(f"  {labels[i]} = {correct[i]}  {status}")
```

### Practice Task 3 — SOLUTION

```python
my_answers = [
    True,    # True or True = True (at least one is True)
    True,    # True or False = True (first one is True)
    True,    # False or True = True (second one is True)
    False    # False or False = False (neither is True)
]

correct = [True or True, True or False, False or True, False or False]
labels = ["True  or True ", "True  or False", "False or True ", "False or False"]

for i in range(4):
    status = "CORRECT" if my_answers[i] == correct[i] else "WRONG"
    print(f"  {labels[i]} = {correct[i]}  {status}")
```

## Section 4 Example: NOT Operator

```python
print(f"not True  = {not True}")
print(f"not False = {not False}")

is_raining = True
if not is_raining:
    print("Let's go outside!")
else:
    print("Better stay in.")

is_raining = False
if not is_raining:
    print("Now let's go outside!")
```

### Practice Task 4 — SOLUTION

```python
print(not True)       # False — flipped from True
print(not False)      # True — flipped from False
print(not (5 > 3))    # False — 5>3 is True, not True is False
print(not (10 == 20)) # True — 10==20 is False, not False is True
```

## Section 5 Example: Order of Operations

```python
step1 = not True
step2 = False and True
step3 = step1 or step2
print(f"Step 1: not True = {step1}")
print(f"Step 2: False and True = {step2}")
print(f"Step 3: {step1} or {step2} = {step3}")
print(f"\nAll at once: not True or False and True = {not True or False and True}")

print(f"\n--- Parentheses change the result ---")
print(f"not True or False and True     = {not True or False and True}")
print(f"not (True or False) and True   = {not (True or False) and True}")
print(f"not (True or False and True)   = {not (True or False and True)}")
print(f"(not True) or (False and True) = {(not True) or (False and True)}")
```

### Practice Task 5 — SOLUTION

```python
# 1. True and True or False
#   Step 1 (and): True and True = True
#   Step 2 (or):  True or False = True
print(f"1. True and True or False = {True and True or False}")  # True

# 2. not False and True
#   Step 1 (not): not False = True
#   Step 2 (and): True and True = True
print(f"2. not False and True = {not False and True}")  # True

# 3. False or False or True
#   Step 1 (or): False or False = False
#   Step 2 (or): False or True = True
print(f"3. False or False or True = {False or False or True}")  # True

# 4. not (True and False)
#   Step 1 (parentheses first): True and False = False
#   Step 2 (not): not False = True
print(f"4. not (True and False) = {not (True and False)}")  # True

# 5. True and not False or False
#   Step 1 (not): not False = True
#   Step 2 (and): True and True = True
#   Step 3 (or):  True or False = True
print(f"5. True and not False or False = {True and not False or False}")  # True
```

## Section 6 Example: Generating Truth Tables

```python
print(f"{'A':<8} {'B':<8} {'C':<8} {'A and (B or not C)':<20}")
print("-" * 44)

for A in [True, False]:
    for B in [True, False]:
        for C in [True, False]:
            result = A and (B or not C)
            print(f"{str(A):<8} {str(B):<8} {str(C):<8} {str(result):<20}")
```

### Practice Task 6 — SOLUTION: XOR Truth Table

```python
# XOR: (A or B) and not (A and B)
# True when A and B are DIFFERENT

print(f"{'A':<8} {'B':<8} {'(A or B) and not (A and B)':<30}")
print("-" * 46)

for A in [True, False]:              # only 2 variables = only 2 nested loops
    for B in [True, False]:
        result = (A or B) and not (A and B)   # the XOR expression
        print(f"{str(A):<8} {str(B):<8} {str(result):<30}")

# Result:
# T, T -> False (same = False)
# T, F -> True  (different = True)
# F, T -> True  (different = True)
# F, F -> False (same = False)
```

## Section 7 Example: Real-World Conditions

```python
age = 16
has_permit = True
has_license = False
is_with_parent = True

can_drive = has_license or (has_permit and is_with_parent)
print(f"Can drive? {can_drive}")

must_take_ed = age < 18 and not has_license
print(f"Must take driver's ed? {must_take_ed}")

can_drive_alone = has_license and age >= 16 and not is_with_parent
print(f"Can drive alone? {can_drive_alone}")
```

### Practice Task 7 — SOLUTION

```python
temperature = 72
is_sunny = True
is_weekend = False
homework_done = True

# 1. Can go to the beach? sunny AND temperature above 75
can_beach = is_sunny and temperature > 75
# True and (72 > 75) = True and False = False
print(f"Can go to beach? {can_beach}")      # False — temp is only 72

# 2. Can play video games? homework done AND (weekend OR temp below 60)
can_play = homework_done and (is_weekend or temperature < 60)
# True and (False or False) = True and False = False
print(f"Can play games? {can_play}")         # False — not weekend and not cold

# 3. Should bring a jacket? NOT sunny OR temp below 65
need_jacket = not is_sunny or temperature < 65
# not True or (72 < 65) = False or False = False
print(f"Need jacket? {need_jacket}")         # False — it's sunny and warm
```
