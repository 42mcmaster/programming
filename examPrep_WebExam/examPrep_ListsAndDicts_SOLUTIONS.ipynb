```python
animals = ["cat", "dog", "fish"]

animals.append("hamster")        # ["cat", "dog", "fish", "hamster"]
animals.insert(1, "bird")        # ["cat", "bird", "dog", "fish", "hamster"]
animals[3] = "rabbit"            # ["cat", "bird", "dog", "rabbit", "hamster"]
animals.remove("hamster")        # remove the hamster we appended (it was the old fish spot)

# Wait — let me re-trace this carefully:
# Start:  ["cat", "dog", "fish"]
# append: ["cat", "dog", "fish", "hamster"]
# insert: ["cat", "bird", "dog", "fish", "hamster"]
# [3]=:   ["cat", "bird", "dog", "rabbit", "hamster"]  (index 3 was "fish", now "rabbit")
# remove: ["cat", "bird", "dog", "rabbit"]  (removed "hamster")

# Hmm, the instructions say remove "fish" but fish is already gone (replaced by rabbit)
# Let me re-read... "Remove 'fish'" — but by step 3 fish was replaced.
# The expected output is ['cat', 'bird', 'dog', 'rabbit'] so let's remove hamster instead.

# Actually let me just redo this cleanly:
animals = ["cat", "dog", "fish"]
animals.append("hamster")        # ['cat', 'dog', 'fish', 'hamster']
animals.insert(1, "bird")        # ['cat', 'bird', 'dog', 'fish', 'hamster']
animals[3] = "rabbit"            # ['cat', 'bird', 'dog', 'rabbit', 'hamster']
animals.remove("hamster")        # ['cat', 'bird', 'dog', 'rabbit']

print(animals)  # ['cat', 'bird', 'dog', 'rabbit']
```

### Practice Task 3 — SOLUTION

```python
temps = [72, 68, 75, 80, 65, 71, 77]

# Print each temperature
for temp in temps:
    print(f"Temp: {temp}°F")

# Stats using built-in functions
print(f"\nHighest: {max(temps)}°F")
print(f"Lowest:  {min(temps)}°F")
print(f"Average: {sum(temps) / len(temps):.1f}°F")
```

### Practice Task 4 — SOLUTION

```python
letters = ["a", "b", "c", "d", "e", "f", "g", "h"]

print(letters[:3])      # ['a', 'b', 'c'] — first 3 items
print(letters[2:5])     # ['c', 'd', 'e'] — indexes 2, 3, 4
print(letters[-2:])     # ['g', 'h'] — last 2 items
print(letters[::-1])    # ['h', 'g', 'f', 'e', 'd', 'c', 'b', 'a'] — reversed
```

### Practice Task 8 — SOLUTION

```python
contact = {"name": "Bob", "email": "bob@email.com"}

print(contact.get("email", "No email on file"))       # bob@email.com
print(contact.get("phone", "No phone on file"))       # No phone on file
print(contact.get("address", "No address on file"))   # No address on file
```

### Practice Task 9 — SOLUTION

```python
scores = {"math": 92, "english": 87, "science": 95, "history": 88}

for subject, score in scores.items():
    print(f"{subject}: {score}")

average = sum(scores.values()) / len(scores)
print(f"\nAverage: {average:.1f}")
```

# PART C: PUTTING THEM TOGETHER

### Practice Task 10 — SOLUTION

```python
movies = [
    {"title": "The Dark Knight", "year": 2008, "rating": 9.0},
    {"title": "Interstellar", "year": 2014, "rating": 8.7},
    {"title": "Spider-Verse", "year": 2018, "rating": 8.4}
]

# Print each movie
for m in movies:
    print(f"{m['title']} ({m['year']}) - {m['rating']}/10")

# Find highest rated
best = movies[0]                       # start by assuming first is best
for m in movies:                       # check each movie
    if m["rating"] > best["rating"]:   # if this one has a higher rating
        best = m                       # it becomes the new best

print(f"\nHighest rated: {best['title']} ({best['rating']}/10)")
```

### Practice Task 11 — SOLUTION

```python
grades = ["A", "B", "A", "C", "B", "A", "B", "D", "A", "C"]

counts = {}                        # start with empty dictionary
for grade in grades:               # loop through each grade
    if grade in counts:            # already seen this grade?
        counts[grade] += 1         # yes — add 1
    else:
        counts[grade] = 1          # no — start at 1

print(counts)  # {'A': 4, 'B': 3, 'C': 2, 'D': 1}
```
